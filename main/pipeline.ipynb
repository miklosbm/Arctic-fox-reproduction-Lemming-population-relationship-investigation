```python
import import_ipynb
from habitat_area_extraction import read_image, resize_mask, apply_mask, hue_transform, calculate_hue_peaks, calculate_habitat_areas
import pandas as pd
from scipy import stats
```

```python
mask_img = read_image("data/valley_orange_polygon_mask.png", grayscale=True)
habitat_img = read_image("data/habitat_type.png", grayscale=False)  # RGB, shape (H, W, 3)

mask_img_resize = resize_mask(mask_img, habitat_img)

masked_habitat_img = apply_mask(habitat_img, mask_img_resize)

habitat_hue_degress = hue_transform(masked_habitat_img)
```

```python
habitat_hue_peaks, habitat_hue_roi = calculate_hue_peaks(habitat_hue_degress, mask_img_resize > 0)
```

```python
peak_labels = ["unknown", "upland", "mesic", "river", "wetland"]  # unknown = mixed color artifact, not a real category
peak_names = dict(zip(habitat_hue_peaks, peak_labels))
```

```python
MAX_HUE_DISTANCE = 16
TOTAL_VALLEY_AREA_KM2 = 65
categ_habitat = calculate_habitat_areas(habitat_hue_roi, habitat_hue_peaks, mask_img_resize, TOTAL_VALLEY_AREA_KM2,
                        peak_names, MAX_HUE_DISTANCE)
```

```python
mesic_area = categ_habitat["mesic"]["area_km2"]
wetland_area = categ_habitat["wetland"]["area_km2"]
```

```python
density_df = pd.read_csv("data/BYLOT-species_density_monitoring.csv")  # load the full monitoring table

lemming_density = density_df[density_df["species_en"].str.contains("lemming", case=False)]  # keep only lemming rows (brown + collared)
lemming_density = lemming_density[lemming_density["habitat"].isin(["mesic", "wetland"])]  # keep only the habitats we have area data for
```

```python
lemming_by_year = lemming_density.pivot_table(
    index=["species_en", "habitat"],   # 4 rows: brown/mesic, brown/wetland, collared/mesic, collared/wetland
    columns="year",                    # one column per year
    values="ind_density_km2"           # the density value that fills the table
)
lemming_by_year
```

```python
habitats_per_row = [idx[1] for idx in lemming_by_year.index]  # ["mesic", "wetland", "mesic", "wetland"]

habitat_area = []
for h in habitats_per_row:
  if h == "mesic":
      habitat_area.append(mesic_area)
  else:
      habitat_area.append(wetland_area)
```

```python
lemming_count_by_year = lemming_by_year.multiply(habitat_area, axis=0)  # density * area = estimated count, for every year
lemming_count_by_year
```

```python
habitat_totals = lemming_count_by_year.groupby(level="habitat").sum()  # brown + collared, per habitat
habitat_totals.index = pd.MultiIndex.from_tuples(
    [("", habitat) for habitat in habitat_totals.index], names=lemming_count_by_year.index.names
)

grand_total = lemming_count_by_year.sum().to_frame().T  # both species + both habitats
grand_total.index = pd.MultiIndex.from_tuples([("", "total")], names=lemming_count_by_year.index.names)

lemming_summary = pd.concat([lemming_count_by_year, habitat_totals, grand_total])
lemming_summary
```

```python
lemming_summary = lemming_count_by_year.copy()

# sum each habitat across both species
lemming_summary.loc[("All", "mesic"), :] = lemming_count_by_year.xs("mesic", level="habitat").sum()
lemming_summary.loc[("All", "wetland"), :] = lemming_count_by_year.xs("wetland", level="habitat").sum()

# sum everything for the grand total
lemming_summary.loc[("All", "mesic+wetland"), :] = lemming_count_by_year.sum()
lemming_summary
```

```python
lemming_summary.to_csv('data/saved/lemming_summary.csv', index=True)
```

```python
natal_dens = pd.read_csv('data/saved/natal_dens_per_year.csv')
```

```python
lem_population = (
    lemming_summary.loc[("All", "mesic+wetland")]
    .rename('Lemming_population')
    .rename_axis('Year')
    .reset_index()
)
print(lem_population)
```

```python
den_lemming_data = pd.merge(natal_dens, lem_population, on='Year', how='inner')
print(den_lemming_data)
```

```python
stat_lemming, p_lemming = stats.shapiro(den_lemming_data['Lemming_population'])
stat_dens, p_dens = stats.shapiro(den_lemming_data['Arctic_fox_natal_dens'])

print(f"Lemming population â€” Shapiro-Wilk: W = {stat_lemming:.3f}, p = {p_lemming:.4f}")
print(f"Arctic fox natal dens â€” Shapiro-Wilk: W = {stat_dens:.3f}, p = {p_dens:.4f}")
```

```python
# correlation
r, p = stats.pearsonr(den_lemming_data['Lemming_population'], den_lemming_data['Arctic_fox_natal_dens'])
rho, p_rho = stats.spearmanr(den_lemming_data['Lemming_population'], den_lemming_data['Arctic_fox_natal_dens'])

print(f"Pearson r = {r:.3f}, p = {p:.4f}")
print(f"Spearman rho = {rho:.3f}, p = {p_rho:.4f}")
```

```python
# lag test: shift lemming values by 1 year
lemming_lag1 = den_lemming_data['Lemming_population'].shift(1)

r_lag, p_lag = stats.pearsonr(lemming_lag1.dropna(), den_lemming_data['Arctic_fox_natal_dens'][1:])
rho_lag, p_rho_lag = stats.spearmanr(lemming_lag1.dropna(), den_lemming_data['Arctic_fox_natal_dens'][1:])

print(f"Lagged Pearson r = {r_lag:.3f}, p = {p_lag:.4f}")
print(f"Lagged Spearman rho = {rho_lag:.3f}, p = {p_rho_lag:.4f}")
```
